```python
from pyspark.sql import functions as F

ruta = "/Volumes/vigia/bronze/landing/nino/"
df = spark.read.text(ruta)
df.printSchema()
display(df)
```

```python
display(df.select("fecha_descarga", "value", "_metadata.file_path").limit(10))
```

```python

datos = df.where(F.col("value").rlike(r"^\s*\d{2}[A-Z]{3}\d{4}\s"))
print(datos.count())


display(datos.select("value", F.regexp_extract_all("value", F.lit(r"-?\d+\.\d"), 0).alias("valores")).limit(5))
```
